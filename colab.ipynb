```python
# @title 🌐 2. Launch OmniWhisper (Colab Version)
import os
%cd /content/OmniWhisper

# 1. Model Path Configuration
whisper_dir = '/content/OmniWhisper/whisper-large-v3-turbo'
model_dir = '/content/OmniWhisper/omnivoice/resources'

# 2. Automatic Offline Mode (Only if models exist)
def check_model(path):
    if not os.path.exists(path): return False
    return any(f.endswith(('.bin', '.safetensors', '.pt')) for f in os.listdir(path))

if check_model(whisper_dir) and check_model(model_dir):
    print("✅ Models found locally. Enabling offline mode for stability.")
    os.environ['TRANSFORMERS_OFFLINE'] = '1'
    os.environ['HF_HUB_OFFLINE'] = '1'
else:
    print("⚠️ Models missing or incomplete. Offline mode disabled for download.")
    os.environ['TRANSFORMERS_OFFLINE'] = '0'
    os.environ['HF_HUB_OFFLINE'] = '0'

os.environ['PYTHONPATH'] = '/content/OmniWhisper'

# 3. Run App
!python omnivoice/app_colab.py --model "$model_dir" --whisper "$whisper_dir" --share
```
