```python
# @title 🌐 2. Launch OmniWhisper
# Ensure we are in the project directory
import os
if not os.path.exists('omnivoice'):
    %cd /content/OmniWhisper

# Set environment for local model usage
os.environ['TRANSFORMERS_OFFLINE'] = '1'
os.environ['HF_HUB_OFFLINE'] = '1'
os.environ['PYTHONPATH'] = os.getcwd()

# Pre-cache Whisper if not present
whisper_dir = './whisper-large-v3-turbo'
if not os.path.exists(whisper_dir):
    print("📥 Downloading Whisper model...")
    os.environ['TRANSFORMERS_OFFLINE'] = '0'
    os.environ['HF_HUB_OFFLINE'] = '0'
    from huggingface_hub import snapshot_download
    snapshot_download(
        repo_id='openai/whisper-large-v3-turbo',
        local_dir=whisper_dir,
        local_dir_use_symlinks=False
    )
    os.environ['TRANSFORMERS_OFFLINE'] = '1'
    os.environ['HF_HUB_OFFLINE'] = '1'

!python omnivoice/app.py --model "./omnivoice/resources" --whisper "$whisper_dir" --share
```
