```python
# @title 🚀 1. Setup & Install (Auto-Restart)
import os
from IPython.display import clear_output

if not os.path.exists('/content/OmniWhisper'):
    print("📥 Cloning OmniWhisper...")
    !git clone https://github.com/nohuggy/OmniWhisper.git /content/OmniWhisper

%cd /content/OmniWhisper

if not os.path.exists("/content/.omni_ready"):
    print("📦 Installing dependencies...")
    !pip install -r requirements_colab.txt --quiet
    
    with open("/content/.omni_ready", "w") as f:
        f.write("done")
    
    clear_output()
    print("✅ Installation complete. RESTARTING RUNTIME...")
    print("💡 After the runtime reconnects, please run the 'Launch' cell below.")
    import os
    os.kill(os.getpid(), 9)
else:
    print("✅ Environment ready. Please run the 'Launch' cell below.")
```
